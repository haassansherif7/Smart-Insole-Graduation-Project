```python
pressure_cols = ['BigF_P','Center_P','Side_P','Pres1','Pres2']
df = df[(df[pressure_cols] <= 4095).all(axis=1)].reset_index(drop=True)
```

```python
temp_cols = ['BigF_T','Center_T','Side_T','Temp1','Temp2']
df = df[(df[temp_cols] >= 20).all(axis=1) & (df[temp_cols] <= 40).all(axis=1)].reset_index(drop=True)
```

```python
df[pressure_cols] = df[pressure_cols].clip(lower=0, upper=4095)
df[temp_cols] = df[temp_cols].clip(lower=20, upper=40)
```

```python
df
```

```python
df_sick = df.copy()    # داتا المرضى فقط (انت شغال عليها أصلاً)

target_size = 25000
current_size = len(df_sick)

extra_needed = target_size - current_size

print("Current sick rows:", current_size)
print("Extra needed:", extra_needed)

def add_noise(df, std_factor=0.015):
    noisy = df.copy()
    for col in df.columns:
        if col != "Label":
            noise = np.random.normal(0, df[col].std() * std_factor, size=len(df))
            noisy[col] = df[col] + noise
    return noisy

if extra_needed > 0:
    # نخليها random sample بس هنعمل عليها Noise عشان تبقى جديدة
    new_rows_raw = df_sick.sample(extra_needed, replace=True, random_state=42)
    new_rows = add_noise(new_rows_raw)
    
    df = pd.concat([df_sick, new_rows], ignore_index=True)
else:
    df = df_sick.sample(target_size, random_state=42)

df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Final sick rows:", len(df))
print("Duplicated rows:", df.duplicated().sum())
```

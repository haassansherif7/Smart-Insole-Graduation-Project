```python
cols_negative = ['Triglycerides','Physical Activity','Diet Score','Stress Level']

for col in cols_negative:
    median = df[df[col] >= 0][col].median()     
    df.loc[df[col] < 0, col] = median
```

```python
for col in ['Age','Glucose','Blood Pressure']:
    df[col].fillna(df[col].median(), inplace=True)
for col in ['Gender']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
def check_all_outliers(df):
    for col in df.select_dtypes(include=['int64','float64']).columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        count = df[(df[col] < lower) | (df[col] > upper)][col].count()
        print(f"{col}: {count} outliers")
check_all_outliers(df)
```

```python
def winsorize_column(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df[col] = np.where(df[col] < lower, lower,
              np.where(df[col] > upper, upper, df[col]))
    return df

numeric_cols = df.select_dtypes(include=['float64','int64']).columns

for col in numeric_cols:
    df = winsorize_column(df, col)

print("✔ Outliers capped successfully.")
```

```python
df_clean = df[df["Medical Condition"].notna()].copy()
df_missing = df[df["Medical Condition"].isna()].copy()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_clean["Medical Condition"] = le.fit_transform(df_clean["Medical Condition"])

print("Classes mapping:", dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
x=df_clean.drop('Medical Condition',axis=1)
y=df_clean['Medical Condition']
```

```python
categorical_cols = ["Gender"]
numeric_cols = [col for col in x.columns if col not in categorical_cols]

print("Features:", x.columns.tolist())
print("Categorical:", categorical_cols)
print("Numeric:", numeric_cols)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Step 2: Preprocessing Transformer (Encoding + Scaling)
preprocess = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_cols)
    ]
)

print("Preprocessing Block Ready ✔")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Step 3: Build Full Pipeline (Preprocessing + Model)
model = Pipeline([
    ("preprocess", preprocess),
    ("clf", RandomForestClassifier(
        n_estimators=400,
        random_state=42,
        n_jobs=-1
    ))
])

print("Pipeline Created ✔")
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Predict on test set
y_pred = model.predict(x_test)

# Print evaluation results
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
from xgboost import XGBClassifier

xgb_model = Pipeline([
    ("preprocess", preprocess),
    ("clf", XGBClassifier(
        n_estimators=600,
        learning_rate=0.03,
        max_depth=6,
        subsample=0.9,
        colsample_bytree=0.9,
        gamma=0.4,
        reg_lambda=2,
        min_child_weight=4,     # مهم جداً لتحسين الفئة الصغيرة
        random_state=42,
        n_jobs=-1,
        eval_metric="mlogloss"
        ))
])

xgb_model.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Predict on test set
y_predd = xgb_model.predict(x_test)

# Print evaluation results
print("Accuracy:", accuracy_score(y_test, y_predd))
print("\nClassification Report:\n", classification_report(y_test, y_predd))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_predd))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


# 2) Create comparison table
#===============================
comparison = pd.DataFrame({
    "Actual": le.inverse_transform(y_test),
    "Predicted": le.inverse_transform(y_predd)
})

#===============================
# 3) Add Match Column
#===============================
comparison["Match"] = comparison["Actual"] == comparison["Predicted"]

print("\n======= SAMPLE RESULTS =======")
print(comparison.head(20))               # أول 20 سطر
print("\nTotal Test Samples:", len(comparison))
print("Correct Predictions:", comparison['Match'].sum())
print("Wrong Predictions:", len(comparison)-comparison['Match'].sum())
print(f"Accuracy from comparison: {comparison['Match'].mean()*100:.2f}%")

#===============================
# 4) Show only wrong predictions
#===============================
wrong = comparison[comparison["Match"] == False]
print("\n======= Wrong Predictions Samples =======")
print(wrong.head(15))

#===============================
# 5) Confusion Table
#===============================
print("\n======= Confusion Table =======")
conf_table = pd.crosstab(comparison["Actual"], comparison["Predicted"])
print(conf_table)

#===============================
# 6) Heatmap Visualization
#===============================
plt.figure(figsize=(8,6))
sns.heatmap(confusion_matrix(y_test, y_predd), annot=True, fmt="d",
            xticklabels=le.classes_, yticklabels=le.classes_,
            cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix Heatmap")
plt.show()
```
